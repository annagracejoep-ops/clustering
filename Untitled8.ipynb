```python
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
display(iris_df)
print(iris_df.info())
```

```python
#KMeans is an iterative clustering algorithm that aims to partition n observations into k clusters, where each observation belongs to the cluster with the nearest mean (centroid).
# Randomly select k data points as initial cluster centroids.
# Assign each data point to the closest centroid, forming k clusters.
# Recalculate the centroids as the mean of all data points assigned to that cluster.
#Repeat the assignment and update steps until the centroids no longer change significantly, or a maximum number of iterations is reached.
```

```python
#The Iris dataset consists of four numerical features (sepal length, sepal width, petal length, petal width).
#KMeans works well with numerical data where the concept of a 'mean' or 'centroid' is meaningful.
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', max_iter=300, n_init=10, random_state=42)
    kmeans.fit(iris_df)
    wcss.append(kmeans.inertia_)


plt.figure(figsize=(10, 6))
plt.plot(range(1, 11), wcss, marker='o', linestyle='--')
plt.title('Elbow Method For Optimal k')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Within-Cluster Sum of Squares (WCSS)')
plt.xticks(range(1, 11))
plt.grid(True)
plt.show()


k = 3
kmeans_model = KMeans(n_clusters=k, init='k-means++', max_iter=300, n_init=10, random_state=42)
clusters = kmeans_model.fit_predict(iris_df)
iris_df['kmeans_cluster'] = clusters

plt.figure(figsize=(12, 8))
sns.scatterplot(
    x='sepal length (cm)',
    y='petal length (cm)',
    hue='kmeans_cluster',
    palette='viridis',
    data=iris_df,
    s=100,
    alpha=0.8
)
plt.scatter(
    kmeans_model.cluster_centers_[:, iris_df.columns.get_loc('sepal length (cm)')],
    kmeans_model.cluster_centers_[:, iris_df.columns.get_loc('petal length (cm)')],
    s=300,
    c='red',
    marker='X',
    label='Centroids'
)
plt.title(f'KMeans Clustering (k={k}) on Iris Dataset')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')
plt.legend()
plt.grid(True)
plt.show()


from sklearn.decomposition import PCA

pca = PCA(n_components=2)
iris_pca = pca.fit_transform(iris_df.drop('kmeans_cluster', axis=1))
iris_pca_df = pd.DataFrame(data=iris_pca, columns=['Principal Component 1', 'Principal Component 2'])
iris_pca_df['kmeans_cluster'] = clusters

plt.figure(figsize=(12, 8))
sns.scatterplot(
    x='Principal Component 1',
    y='Principal Component 2',
    hue='kmeans_cluster',
    palette='viridis',
    data=iris_pca_df,
    s=100,
    alpha=0.8
)
plt.scatter(
    pca.transform(kmeans_model.cluster_centers_)[:, 0],
    pca.transform(kmeans_model.cluster_centers_)[:, 1],
    s=300,
    c='red',
    marker='X',
    label='Centroids (Projected)'
)
plt.title(f'KMeans Clustering (k={k}) on Iris Dataset (PCA Reduced)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Hierarchical clustering (also known as Hierarchical Cluster Analysis or HCA) is a method of cluster analysis which seeks to build a hierarchy of clusters.
#It starts with each data point as its own cluster, and then iteratively merges the two closest clusters until only one cluster (or a predefined number of clusters) remains.
```

```python
#Unlike KMeans, hierarchical clustering does not require the number of clusters (k) to be specified beforehand. The dendrogram allows us to visually inspect the data and decide on an appropriate number of clusters by cutting the dendrogram at a certain height.
# Like KMeans, hierarchical clustering works effectively with numerical features, which the Iris dataset exclusively comprises.
```

```python
from scipy.cluster.hierarchy import dendrogram, linkage, fcluster
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

scaler = StandardScaler()
iris_scaled = scaler.fit_transform(iris_df.drop('kmeans_cluster', axis=1))


linked = linkage(iris_scaled, method='ward')


plt.figure(figsize=(15, 10))
dendrogram(linked,
           orientation='top',
           distance_sort='descending',
           show_leaf_counts=True)
plt.title('Hierarchical Clustering Dendrogram (Iris Dataset)')
plt.xlabel('Sample Index')
plt.ylabel('Distance')
plt.show()


k_hierarchical = 3
clusters_hierarchical = fcluster(linked, k_hierarchical, criterion='maxclust')


iris_df_hierarchical = iris_df.drop('kmeans_cluster', axis=1).copy()
iris_df_hierarchical['hierarchical_cluster'] = clusters_hierarchical


plt.figure(figsize=(12, 8))
sns.scatterplot(
    x='sepal length (cm)',
    y='petal length (cm)',
    hue='hierarchical_cluster',
    palette='viridis',
    data=iris_df_hierarchical,
    s=100,
    alpha=0.8
)
plt.title(f'Hierarchical Clustering (k={k_hierarchical}) on Iris Dataset')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')
plt.legend(title='Hierarchical Cluster')
plt.grid(True)
plt.show()


from sklearn.decomposition import PCA

pca_hierarchical = PCA(n_components=2)
iris_pca_hierarchical = pca_hierarchical.fit_transform(iris_df_hierarchical.drop('hierarchical_cluster', axis=1))
iris_pca_hierarchical_df = pd.DataFrame(data=iris_pca_hierarchical, columns=['Principal Component 1', 'Principal Component 2'])
iris_pca_hierarchical_df['hierarchical_cluster'] = clusters_hierarchical

plt.figure(figsize=(12, 8))
sns.scatterplot(
    x='Principal Component 1',
    y='Principal Component 2',
    hue='hierarchical_cluster',
    palette='viridis',
    data=iris_pca_hierarchical_df,
    s=100,
    alpha=0.8
)
plt.title(f'Hierarchical Clustering (k={k_hierarchical}) on Iris Dataset (PCA Reduced)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend(title='Hierarchical Cluster')
plt.grid(True)
plt.show()
```
